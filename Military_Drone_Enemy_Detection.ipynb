```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV dosyasını oku
csv_file = "/content/yolov5/datasets/runs/detect/train2/results.csv"
df = pd.read_csv(csv_file)

# Grafik çizme fonksiyonu
def plot_graph(x, y_list, labels, title, xlabel, ylabel):
    plt.figure(figsize=(10, 5))
    for y, label in zip(y_list, labels):
        plt.plot(df[x], df[y], marker='o', linestyle='-', label=label)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend()
    plt.grid()
    plt.show()

# Eğitim ve Validasyon Kayıpları
plot_graph("epoch", ["train/box_loss", "train/cls_loss", "train/dfl_loss"],
           ["Box Loss", "Class Loss", "DFL Loss"], "Train Loss Over Epochs", "Epoch", "Loss")

plot_graph("epoch", ["val/box_loss", "val/cls_loss", "val/dfl_loss"],
           ["Validation Box Loss", "Validation Class Loss", "Validation DFL Loss"],
           "Validation Loss Over Epochs", "Epoch", "Loss")

# Model Performansı (Precision, Recall, mAP)
plot_graph("epoch", ["metrics/precision(B)", "metrics/recall(B)"],
           ["Precision", "Recall"], "Precision & Recall Over Epochs", "Epoch", "Score")

plot_graph("epoch", ["metrics/mAP50(B)", "metrics/mAP50-95(B)"],
           ["mAP50", "mAP50-95"], "mAP Over Epochs", "Epoch", "mAP Score")
```

```python
import cv2
from google.colab.patches import cv2_imshow
import random

# Video dosyasını aç
video_path = "/content/uzun_online.avi"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Hata: Video dosyası açılamadı.")
    exit()

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
random_frames = random.sample(range(total_frames), 10)

# Frameleri gösterme döngüsü
for frame_number in random_frames:
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_number)
    ret, frame = cap.read()

    if not ret:
        print(f"Frame {frame_number} okunamadı.")
        continue

    # Frame'i göster
    cv2_imshow(frame)
    print(f"Frame {frame_number}")

    if cv2.waitKey(0) & 0xFF == ord('q'):
        break


cap.release()
cv2.destroyAllWindows()
```
